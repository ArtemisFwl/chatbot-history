```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
groq_api_key=os.getenv("groq_api_key")
```

```python
from langchain_groq import ChatGroq
model=ChatGroq(model="llama-3.1-8b-instant", groq_api_key=groq_api_key)
model
```

```python
from langchain_core.messages import HumanMessage
model.invoke([HumanMessage(content="Hi, I am aman Deep. I am officer at Indian oil Corp Ltd")])
```

```python
from langchain_core.messages import AIMessage 

model.invoke(
    [
        HumanMessage(content="Hi, I am aman Deep. I am officer at Indian oil Corp Ltd"),
        AIMessage(content="Nice to meet you, Officer Aman Deep from Indian Oil Corp Ltd. How can I assist you today? Are you looking for information related to the oil and gas industry or perhaps something else?"),
        HumanMessage(content="What is my name and where do I work and what is my role.")
    ]
)
```

```python
#Message History
from langchain_community .chat_message_histories import ChatMessageHistory
from langchain_core.chat_history import BaseChatMessageHistory 
from langchain_core.runnables.history import RunnableWithMessageHistory
```

```python
store={} #A new variable that is dictonary

def get_session_history(session_id:str)->BaseChatMessageHistory:
    if session_id not in store:
        store[session_id]=ChatMessageHistory()
    return store[session_id]

with_message_history=RunnableWithMessageHistory(model, get_session_history)
```

```python
config={"configurable":{"session_id":"chat1"}}
```

```python
response=with_message_history.invoke(
    [HumanMessage(content="Hi, I am aman Deep. I am officer at Indian oil Corp Ltd")],
    config=config
)
```

```python
response.content
```

```python
##Change the config---> Changing the session ID

config1={"configurable":{"session_id":"chat1"}}
```

```python
response=with_message_history.invoke(
    [HumanMessage(content="What is my name and what do i do?")],
    config=config1
)
```

```python
response.content
```

## Prompt template

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
prompt=ChatPromptTemplate.from_messages(
    [
        ("system","you are a helpful assistant. Answer all questions to the best of your ability"),
        MessagesPlaceholder(variable_name="messages")
    ]
)

chain=prompt|model
```

```python
chain.invoke({"messages":[HumanMessage(content="Hi My name is Aman Deep")]})
```

```python
config={"configurable":{"session_id":"chat3"}}
```

```python
response=with_message_history.invoke(
    [HumanMessage(content="Hi My name is Aman Deep")],
    config=config)

response.content
```

```python
# Add more complexity

prompt=ChatPromptTemplate.from_messages(
    [
        ("system","you are a helpful assistant. Answer all questions to the best of your ability in {language}"),
        MessagesPlaceholder(variable_name="messages")
    ]
)

chain=prompt|model

```

```python
response=chain.invoke({"messages":[HumanMessage(content="Hi I am Aman Deep")],
                       "language":"Hindi"})
response.content
```

```python
with_message_history=RunnableWithMessageHistory(
    chain,
    get_session_history,
    input_messages_key="messages"
)
```

```python
config={"configurable":{"session_id":"chat4"}}
```

```python
response=with_message_history.invoke(
    {"messages":[HumanMessage(content="Hi My name is Aman Deep")],
     "language":"Hindi"},
    config=config)

response.content
```
